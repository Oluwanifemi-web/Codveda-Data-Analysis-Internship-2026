```python
import pandas as pd
df = pd.read_csv("1) iris.csv")
```

```python
print("First 5 rows are:")
print(df.head())
```

```python
print("The size of the dataset is:")
print(df.shape)
```

```python
print("The column names and data types is:")
print(df.info())
```

```python
#Check for any missing values across all columns
print("Missing vales per column:")
print(df.isnull().sum())
```

```python
#Check for duplicate entries
print("Number of duplicated rows:")
print(df.duplicated().sum())
```

```python
#Drop duplicates if any exist, and update the dataset
if df.duplicated().sum()>0:
    df = df.drop_duplicates()
    print(f"Duplicates dropped. New dataset shape: {df.shape}")
else:
    print("No duplicates found. Dataset is already cleaned.")
```

```python
df.to_csv("level_1", index=False)
print("Cleaned Dataset successfully")
```
