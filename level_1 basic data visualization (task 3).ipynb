```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#To load our cleaned dataset
df = pd.read_csv("level_1 data cleaned.csv")
```

```python
#Let's set a cleaned visual theme
sns.set_theme(style="whitegrid")
```

```python
#Scatter Plot: Petal Length vs Width
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df, x="petal_length", y="petal_width", hue="species")
plt.title("Petal Length vs Width by Species")
plt.savefig("iris_scatter.png")
plt.show()

#Bar Plot: Average Sepal Length per Species
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x="species", y="sepal_length", errorbar=None)
plt.title("Average Sepal Length per Species")
plt.savefig("iris_bar.png")
plt.show()

#Line Chart: Sepal Width Trend
plt.figure(figsize=(6, 3))
sns.lineplot(data=df.head(50), y="sepal_width", x=df.head(50).index)
plt.title("Sepal Width Trend (First 50 Samples)")
plt.savefig("iris_line.png")
plt.show()
```
